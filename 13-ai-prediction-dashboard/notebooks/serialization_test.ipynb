```python
import pickle
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

print("Imports done! ✅")
```

```python
# Load pkl files
with open('engineered_model.pkl', 'rb') as f:
    model = pickle.load(f)

with open('engineered_features.pkl', 'rb') as f:
    feature_names = pickle.load(f)

print("Model loaded! ✅")
print("Total features expected:", len(feature_names))
print("\nFirst 5 feature names:")
print(feature_names[:5])
```

```python
#  See all feature names
for i, feature in enumerate(feature_names):
    print(f"{i+1}. {feature}")
```

```python
# Create dummy customer + predict
import numpy as np

# One row with 45 zeros
dummy = pd.DataFrame([np.zeros(len(feature_names))], columns=feature_names)

# Predict
prob = model.predict_proba(dummy)[0][1]
print(f"Churn probability: {prob:.4f}")
print(f"Churn probability %: {prob*100:.2f}%")
```

```python
# High risk customer
high_risk = pd.DataFrame([np.zeros(len(feature_names))], columns=feature_names)

# Set high risk signals
high_risk['ContractRiskScore'] = 3
high_risk['IsNewCustomer'] = 1
high_risk['FiberAndMonthly'] = 1
high_risk['PriceShockFeature'] = 1
high_risk['tenure'] = 2
high_risk['MonthlyCharges'] = 85

prob_high = model.predict_proba(high_risk)[0][1]
print(f"High risk customer churn probability: {prob_high*100:.2f}%")
```

```python
low_risk = pd.DataFrame([np.zeros(len(feature_names))], columns=feature_names)

low_risk['ContractRiskScore'] = 1
low_risk['IsLongTermCustomer'] = 1
low_risk['IsAutoPayment'] = 1
low_risk['tenure'] = 58
low_risk['MonthlyCharges'] = 45
low_risk['Contract_Two year'] = 1

prob_low = model.predict_proba(low_risk)[0][1]
print(f"Low risk customer churn probability: {prob_low*100:.2f}%")
```
